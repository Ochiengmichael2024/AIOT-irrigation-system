# Data Preprocessing
Load a raw sensor CSV, inspect missing values, apply the project's short-gap interpolation helper, and save the cleaned dataset.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
from src.preprocessing import handle_missing_sensor_data, load_raw_data, save_processed_data
```

```python
raw_files = sorted((PROJECT_ROOT / 'data' / 'raw').glob('*.csv'))
if not raw_files:
    raise FileNotFoundError('Add a CSV dataset to data/raw/ before continuing.')
raw_data = load_raw_data(raw_files[0])
display(raw_data.isna().sum())
clean_data = handle_missing_sensor_data(raw_data)
output_path = save_processed_data(clean_data, f'{raw_files[0].stem}_processed.csv')
print(f'Saved {len(clean_data)} rows to {output_path}')
```
