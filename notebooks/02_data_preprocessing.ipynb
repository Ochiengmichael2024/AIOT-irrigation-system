```python
from pathlib import Path
import pandas as pd

# Locate the project root
PROJECT_ROOT = (
    Path.cwd().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd()
)

# Locate the raw data directory
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

# Find CSV files
raw_files = sorted(RAW_DATA_DIR.glob("*.csv"))

# Check whether any CSV files exist
if not raw_files:
    raise FileNotFoundError(
        f"No CSV files found in {RAW_DATA_DIR}"
    )

# Load the first CSV file
raw_data = pd.read_csv(raw_files[0])

# Display basic information
print("Loaded file:", raw_files[0].name)
print("Raw dataset shape:", raw_data.shape)
print("Columns:", raw_data.columns.tolist())

raw_data.head()
```

```python
clean_data = handle_missing_sensor_data(raw_data)
print("Processed dataset shape:", clean_data.shape)
print("Missing values:", clean_data.isna().sum().sum())
```

```python
output_path = save_processed_data(
    clean_data,
    f"{raw_files[0].stem}_processed.csv"
)

print(f"Saved {len(clean_data)} rows to {output_path}")
```

```python
print("Duplicate rows:", clean_data.duplicated().sum())

print("\nData types:")
print(clean_data.dtypes)

print("\nStatistical summary:")
display(clean_data.describe(include="all").T)
```
